### Data Loading

In this section, we import the required Python library and load all datasets into Pandas DataFrames for further analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
orders = pd.read_csv("../data/raw/orders.csv")
order_items = pd.read_csv("../data/raw/order_items.csv")
order_item_refunds = pd.read_csv("../data/raw/order_item_refunds.csv")
website_sessions = pd.read_csv("../data/raw/website_sessions.csv")
website_pageviews = pd.read_csv("../data/raw/website_pageviews.csv")
products = pd.read_csv("../data/raw/products.csv")
```

### Organizing DataFrames

To avoid repeating code for each dataset, all DataFrames are stored in a dictionary. This allows us to iterate through every table efficiently during data exploration.

```python
tables = {
    "Products": products,
    "Orders": orders,
    "Order Items": order_items,
    "Refunds": order_item_refunds,
    "Website Sessions": website_sessions,
    "Website Pageviews": website_pageviews
}
```

### Dataset Overview

##### Objective

Understand the structure of every dataset by examining:

- Number of rows and columns
- Column names
- Data types
- Missing values

This helps identify data quality issues before analysis.

```python
for name, df in tables.items():
    print("=" * 50)
    print(name)
    # shape
    print("Shape:", df.shape)
    # columns
    print("Columns:", list(df.columns))
    # dtypes
    print("Data Types:")
    print(df.dtypes)
    # Missing values
    print("Missing Values:")
    print(df.isnull().sum())
    print()
```

### Data Preview

##### Objective

Inspect the first and last few records of every dataset to understand the stored information, verify data consistency, and identify any unusual values.

```python
for name, df in tables.items():
    print("=" * 50)
    print(name)
    # head
    display(df.head())
    # tail
    display(df.tail())
    print()
```

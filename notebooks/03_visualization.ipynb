```python
# 4.4.5 Orders by Week
# Why Line Chart + Moving Average?
# Weekly data fluctuates a lot. A 4-week moving average highlights the overall trend.

weekly_orders["created_at"] = pd.to_datetime(weekly_orders["created_at"])

# Calculate 4-Week Moving Average
weekly_orders["Orders_MA"] = (
    weekly_orders["Orders"]
    .rolling(window=4)
    .mean()
)

plt.figure(figsize=(12,5))

# Actual Weekly Orders
plt.plot(
    weekly_orders["created_at"],
    weekly_orders["Orders"],
    alpha=0.4,
    label="Weekly Orders"
)

# Trend Line
plt.plot(
    weekly_orders["created_at"],
    weekly_orders["Orders_MA"],
    linewidth=3,
    label="4-Week Moving Average"
)

plt.title("Weekly Orders")
plt.xlabel("Week")
plt.ylabel("Orders")

plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# 4.4.6 Revenue by Week
# Why Line Chart + Moving Average?
# Smooths weekly revenue fluctuations to reveal the overall business trend.

weekly_revenue["created_at"] = pd.to_datetime(weekly_revenue["created_at"])

# Calculate 4-Week Moving Average
weekly_revenue["Revenue_MA"] = (
    weekly_revenue["Revenue"]
    .rolling(window=4)
    .mean()
)

plt.figure(figsize=(12,5))

# Actual Weekly Revenue
plt.plot(
    weekly_revenue["created_at"],
    weekly_revenue["Revenue"],
    alpha=0.4,
    label="Weekly Revenue"
)

# Trend Line
plt.plot(
    weekly_revenue["created_at"],
    weekly_revenue["Revenue_MA"],
    linewidth=3,
    label="4-Week Moving Average"
)

plt.title("Weekly Revenue")
plt.xlabel("Week")
plt.ylabel("Revenue (USD)")

plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# 4.4.7 Daily Orders (Last 3 Months)

daily_orders["Date"] = pd.to_datetime(daily_orders["Date"])

last_date = daily_orders["Date"].max()

daily_orders_3m = daily_orders[
    daily_orders["Date"] >= last_date - pd.DateOffset(months=3)
]

plt.figure(figsize=(12,5))

plt.fill_between(
    daily_orders_3m["Date"],
    daily_orders_3m["Orders"],
    alpha=0.35
)

plt.plot(
    daily_orders_3m["Date"],
    daily_orders_3m["Orders"],
    linewidth=2
)

plt.title("Daily Orders (Last 3 Months)")
plt.xlabel("Date")
plt.ylabel("Orders")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# 4.4.8 Daily Revenue (Last 3 Months)
# Why Area Chart?
# Highlights recent daily revenue trends while making the chart easier to read.

daily_revenue["Date"] = pd.to_datetime(daily_revenue["Date"])

last_date = daily_revenue["Date"].max()

daily_revenue_3m = daily_revenue[
    daily_revenue["Date"] >= last_date - pd.DateOffset(months=3)
]

plt.figure(figsize=(12,5))

plt.fill_between(
    daily_revenue_3m["Date"],
    daily_revenue_3m["Revenue"],
    alpha=0.35
)

plt.plot(
    daily_revenue_3m["Date"],
    daily_revenue_3m["Revenue"],
    linewidth=2
)

plt.title("Daily Revenue (Last 3 Months)")
plt.xlabel("Date")
plt.ylabel("Revenue (USD)")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
